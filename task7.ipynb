```python
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"   # Default status

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def display_status(self):
        print(f"{self.name} is {self.status}")



light = SmartLight("Bedroom Light")
light.turn_on()


light.display_status()
```

```python
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)



employee1 = Employee("Rahul", "EMP102", "IT")

employee1.display_id_card()
```

```python
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.phone)



contact1 = Contact("Anita", "9876543210")


contact1.display_contact()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹" + str(self.price))



product1 = Product("Headphones", 2499)


product1.display_price_tag()
```

```python
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_details(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 5")


# 📥 Sample Input
movie1 = Movie("Inception", 4.8)

# 📤 Expected Output
movie1.display_details()
```
